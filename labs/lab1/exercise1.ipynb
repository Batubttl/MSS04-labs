```python
import math

def q_solve(a, b, c):
    D = b**2 - 4*a*c
    if D < 0:
        print('no real solution')
    else:
        x1 = (-b + math.sqrt(D)) / (2*a)
        x2 = (-b - math.sqrt(D)) / (2*a)
        return x1, x2

print(q_solve(1, -3, 2))   # (2.0, 1.0)
print(q_solve(1, 2, 1))    # (-1.0, -1.0)
q_solve(1, 0, 1)           # no real solution
```

```python
# complex solution allowed
# solve ax^2+bx+c=0
import math

def q_solve(a, b, c):
    # compute the discriminant D
    D = b**2 - 4*a*c
    if D >= 0:
        sq_D = math.sqrt(D)
    else:
        sq_D = math.sqrt(-D) * 1j
    x1 = (-b + sq_D) / (2*a)
    x2 = (-b - sq_D) / (2*a)
    return x1, x2

x1, x2 = q_solve(1, 4, 1)
print(x1, x2)    # -0.2679... -3.7320...
x1, x2 = q_solve(1, 1, 1)
print(x1, x2)    # (-0.5+0.866j) (-0.5-0.866j)
```

```python
# Exercise 1d – complex coefficients (cmath)
import cmath

def q_solve(a, b, c):
    D = b**2 - 4*a*c
    x1 = (-b + cmath.sqrt(D)) / (2*a)
    x2 = (-b - cmath.sqrt(D)) / (2*a)
    return x1, x2

print(q_solve(1, -3, 2))          # ((2+0j), (1+0j))
print(q_solve(1, 0, 1))           # (1j, -1j)
print(q_solve(1, -(2+1j), 2j))    # ((2+0j), 1j)  → karmaşık katsayılar
```
