```python
# opening/creating and writing a text file
f = open("test.txt", "w+")
f.write("this is a test")
f.close()
# reading a text file
f = open("test.txt", "r")
print(f.read())
f.close()
```

```python
#CSV- files
import csv
#opening/creating and writing to a csv file
with open("test.csv",mode="w", newline = "") as f :
#create a writer variable that allows to write to the csv-file
writer = csv.writer(f,delimiter = ",")
#write to the CSV-file row by row
writer.writerow(["Name","City"])
writer.writerow(["Bob","London"])
```
